```python
# C:\Users\sagni\Downloads\ReceiptRadar\infer.py
# ReceiptRadar — Inference helpers:
#   - OCR (docTR / easyocr / pytesseract with graceful fallback)
#   - Token feature extraction (matches training heuristics)
#   - Keras model inference (parser_model.h5)
#   - Aggregation to structured JSON (merchant/date/subtotal/tax/total + line items)
#   - Fraud scoring (rules + pHash duplicate suspicion)
#   - Overlay visualization (boxes + predicted labels)

import os, re, io, json, time, base64, pickle, warnings
from pathlib import Path
warnings.filterwarnings("ignore")

import numpy as np
from PIL import Image, ImageDraw, ImageFont

import tensorflow as tf
from tensorflow import keras

# -----------------------------
# Config / Paths
# -----------------------------
BASE_DIR    = r"C:\Users\sagni\Downloads\ReceiptRadar"
MODEL_H5    = os.path.join(BASE_DIR, "parser_model.h5")
PP_PATH     = os.path.join(BASE_DIR, "preprocessor.pkl")
UPLOAD_DIR  = os.path.join(BASE_DIR, "uploads")
OVERLAY_DIR = os.path.join(BASE_DIR, "overlays")
INDEX_DIR   = os.path.join(BASE_DIR, "artifacts")  # pHash index here

os.makedirs(UPLOAD_DIR, exist_ok=True)
os.makedirs(OVERLAY_DIR, exist_ok=True)
os.makedirs(INDEX_DIR, exist_ok=True)

PHASH_NPY   = os.path.join(INDEX_DIR, "phash_index.npy")
PHASH_META  = os.path.join(INDEX_DIR, "phash_meta.csv")

# -----------------------------
# Load model & preprocessor
# -----------------------------
model = keras.models.load_model(MODEL_H5, compile=False)
with open(PP_PATH, "rb") as f:
    preproc = pickle.load(f)

scaler   = preproc["scaler"]
cat2id   = preproc["cat2id"]
id2cat   = preproc["id2cat"]
CATS     = [id2cat[i] for i in range(len(id2cat))]
N_CLASS  = len(CATS)

regex_cfg = preproc.get("regex", {})
DATE_RE   = re.compile(regex_cfg.get("date", r"\b\d{4}[-/]\d{1,2}[-/]\d{1,2}\b"))
PRICE_RE  = re.compile(regex_cfg.get("price", r"^\$?\s*\d{1,4}(\,\d{3})*(\.\d{1,2})?$"), re.IGNORECASE)
TOTAL_KW  = re.compile(regex_cfg.get("total_kw", r"\b(total|amount\s*due)\b"), re.IGNORECASE)
TAX_KW    = re.compile(regex_cfg.get("tax_kw", r"\btax\b"), re.IGNORECASE)
ADDR_KW   = re.compile(regex_cfg.get("addr_kw", r"\b(st|street|ave|road|rd|blvd|dr|suite)\b"), re.IGNORECASE)

# -----------------------------
# OCR setup with graceful fallback
# -----------------------------
OCR_BACKEND = preproc.get("ocr_backend", "none")
_doctr = None
_easy  = None

def _try_doctr():
    try:
        from doctr.models import ocr_predictor
        from doctr.io import DocumentFile
        return ocr_predictor(pretrained=True)
    except Exception:
        return None

def _try_easyocr():
    try:
        import easyocr
        return easyocr.Reader(["en"], gpu=False, verbose=False)
    except Exception:
        return None

def _tesseract_ok():
    try:
        import pytesseract
        default_tess = r"C:\Program Files\Tesseract-OCR\tesseract.exe"
        if os.path.exists(default_tess):
            pytesseract.pytesseract.tesseract_cmd = default_tess
        return True
    except Exception:
        return False

if OCR_BACKEND == "doctr":
    _doctr = _try_doctr()
    if _doctr is None: OCR_BACKEND = "easyocr"
if OCR_BACKEND == "easyocr":
    _easy = _try_easyocr()
    if _easy is None: OCR_BACKEND = "pytesseract"
if OCR_BACKEND == "pytesseract":
    if not _tesseract_ok(): OCR_BACKEND = "none"
if OCR_BACKEND == "none":
    _doctr = _try_doctr()
    if _doctr is not None:
        OCR_BACKEND = "doctr"
    else:
        _easy = _try_easyocr()
        if _easy is not None:
            OCR_BACKEND = "easyocr"
        elif _tesseract_ok():
            OCR_BACKEND = "pytesseract"

# -----------------------------
# OCR wrappers
# -----------------------------
IMG_EXTS = {".jpg",".jpeg",".png",".tif",".tiff",".bmp",".gif",".webp"}
PDF_EXTS = {".pdf"}

def ocr_tokens(path):
    ext = Path(path).suffix.lower()
    if OCR_BACKEND == "doctr" and _doctr is not None:
        try:
            from doctr.io import DocumentFile
            if ext in PDF_EXTS:
                doc = DocumentFile.from_pdf(path)
            else:
                doc = DocumentFile.from_images(path)
            res = _doctr(doc)
            toks=[]
            for p in res.pages:
                for b in p.blocks:
                    for l in b.lines:
                        for w in l.words:
                            txt = (w.value or "").strip()
                            if not txt: continue
                            box = [float(w.geometry[0][0]), float(w.geometry[0][1]),
                                   float(w.geometry[1][0]), float(w.geometry[1][1])]
                            conf = float(getattr(w, "confidence", 0.9) or 0.9)
                            toks.append({"text":txt, "box":box, "conf":conf})
            if toks: return toks
        except Exception:
            pass
    if OCR_BACKEND == "easyocr" and _easy is not None and ext not in PDF_EXTS:
        try:
            res = _easy.readtext(path, detail=1)
            im = Image.open(path).convert("RGB"); W,H = im.size
            toks=[]
            for bbox, txt, conf in res:
                xs = [p[0] for p in bbox]; ys=[p[1] for p in bbox]
                x0,y0,x1,y1 = max(min(xs),0), max(min(ys),0), min(max(xs),W), min(max(ys),H)
                box = [x0/max(W,1), y0/max(H,1), x1/max(W,1), y1/max(H,1)]
                txt = (txt or "").strip()
                if txt:
                    toks.append({"text":txt, "box":box, "conf":float(conf)})
            if toks: return toks
        except Exception:
            pass
    if OCR_BACKEND in ("pytesseract","none") and ext not in PDF_EXTS:
        try:
            import pytesseract
            from pytesseract import Output
            im = Image.open(path).convert("RGB"); W,H = im.size
            data = pytesseract.image_to_data(im, output_type=Output.DICT)
            toks=[]; n=len(data.get("text",[]))
            for i in range(n):
                txt = (data["text"][i] or "").strip()
                if not txt: continue
                conf = data.get("conf",[0]*n)[i]
                try: conf = float(conf) if conf not in ("-1","") else 0.0
                except: conf = 0.0
                x,y = data["left"][i], data["top"][i]
                w,h = data["width"][i], data["height"][i]
                if W>0 and H>0 and w>0 and h>0:
                    box = [x/W, y/H, (x+w)/W, (y+h)/H]
                else:
                    box = [0,0,0,0]
                toks.append({"text":txt, "box":box, "conf":conf/100.0})
            if toks: return toks
        except Exception:
            pass
    # fallback: filename tokens
    base = Path(path).stem.replace("_"," ").replace("-"," ")
    return [{"text":s, "box":[0,0,1,0.1], "conf":0.5} for s in base.split() if s]

# -----------------------------
# Features & pseudo-labels (same logic as training)
# -----------------------------
def token_features(token):
    x0,y0,x1,y1 = token["box"]
    w = max(x1-x0, 1e-6); h = max(y1-y0, 1e-6)
    area = w*h
    cx, cy = 0.5*(x0+x1), 0.5*(y0+y1)
    txt = token["text"]
    L = len(txt)
    digits = sum(ch.isdigit() for ch in txt)
    alphas = sum(ch.isalpha() for ch in txt)
    uppers = sum(ch.isupper() for ch in txt)
    digit_ratio = digits / max(L,1)
    alpha_ratio = alphas / max(L,1)
    upper_ratio = uppers / max(alphas,1)
    price_like = 1.0 if PRICE_RE.search(txt) else 0.0
    date_like  = 1.0 if DATE_RE.search(txt) else 0.0
    tax_kw     = 1.0 if TAX_KW.search(txt) else 0.0
    total_kw   = 1.0 if TOTAL_KW.search(txt) else 0.0
    addr_kw    = 1.0 if ADDR_KW.search(txt) else 0.0
    conf       = float(token.get("conf", 0.5))
    return np.array([x0,y0,x1,y1,w,h,area,cx,cy,L,digit_ratio,alpha_ratio,upper_ratio,
                     price_like,date_like,tax_kw,total_kw,addr_kw,conf], dtype=np.float32)

def predict_tags(tokens):
    if not tokens:
        return []
    X = np.vstack([token_features(t) for t in tokens]).astype(np.float32)
    Xs = scaler.transform(X)
    y_prob = model.predict(Xs, verbose=0)
    y_pred = y_prob.argmax(axis=1)
    return [ (CATS[i], float(y_prob[j, i])) for j,i in enumerate(y_pred) ]

# -----------------------------
# Aggregation: tokens -> structured JSON
# -----------------------------
def _as_float(s):
    try:
        s = s.replace(",", "")
        m = re.search(r"(\d+(?:\.\d{1,2})?)", s)
        if m: return float(m.group(1))
    except Exception:
        pass
    return None

def aggregate_receipt(tokens, preds):
    # group by center y to form lines
    ys = [0.5*(t["box"][1]+t["box"][3]) for t in tokens]
    idxs = list(range(len(tokens)))
    # line clustering by proximity
    lines = []
    used = set()
    for i in idxs:
        if i in used: continue
        cy = ys[i]
        group = [j for j in idxs if abs(ys[j]-cy) < 0.015]
        group.sort(key=lambda j: tokens[j]["box"][0])
        for j in group: used.add(j)
        lines.append(group)
    # choose fields
    merchant = ""
    date     = ""
    subtotal = None
    tax      = None
    total    = None
    items    = []

    # pre-pick totals: find line with any 'total' tag; else bottom-most price-like
    total_line = None
    for g in lines:
        if any(preds[j][0] == "total" for j in g):
            total_line = g
    if total_line is None:
        price_rows = [g for g in lines if any(PRICE_RE.search(tokens[j]["text"] or "") for j in g)]
        if price_rows:
            total_line = price_rows[-1]

    # merchant: top line with alpha-dominant text
    if lines:
        top = lines[0]
        alpha_counts = [(sum(ch.isalpha() for ch in tokens[j]["text"]), j) for j in top]
        alpha_counts.sort(reverse=True)
        if alpha_counts and alpha_counts[0][0] >= 3:
            merchant = tokens[alpha_counts[0][1]]["text"]

    # parse date: first token with date-like
    for i,(lbl,_) in enumerate(preds):
        if lbl == "date" or DATE_RE.search(tokens[i]["text"]):
            date = tokens[i]["text"]; break

    # pick total on total_line
    if total_line:
        prices = []
        for j in total_line:
            v = _as_float(tokens[j]["text"])
            if v is not None: prices.append(v)
        if prices:
            total = max(prices)

    # tax/subtotal: choose best matches on lines that have respective keywords
    for g in lines:
        g_text = " ".join(tokens[j]["text"] for j in g)
        if tax is None and TAX_KW.search(g_text):
            vals = [_as_float(tokens[j]["text"]) for j in g]
            vals = [v for v in vals if v is not None]
            if vals:
                tax = max(vals)
        if subtotal is None and re.search(r"\b(sub\s*total|subtotal)\b", g_text, re.IGNORECASE):
            vals = [_as_float(tokens[j]["text"]) for j in g]
            vals = [v for v in vals if v is not None]
            if vals:
                subtotal = max(vals)

    # line-items: lines with some alpha on left and a price at right above subtotal/total lines
    if lines:
        # consider all lines above total_line as items region
        if total_line:
            total_cy = np.mean([ys[j] for j in total_line])
            candidate_lines = [g for g in lines if np.mean([ys[j] for j in g]) < total_cy]
        else:
            candidate_lines = lines[:-1]

        for g in candidate_lines:
            left_text = " ".join(tokens[j]["text"] for j in g if tokens[j]["box"][0] < 0.6)
            right_vals = [_as_float(tokens[j]["text"]) for j in g if tokens[j]["box"][0] >= 0.6]
            right_vals = [v for v in right_vals if v is not None]
            if left_text.strip() and right_vals:
                items.append({"item": left_text.strip(), "amount": float(max(right_vals))})

    out = {
        "merchant": merchant,
        "date": date,
        "subtotal": subtotal,
        "tax": tax,
        "total": total,
        "currency": None,
        "line_items": items,
        "tokens": [
            {"text": tokens[i]["text"], "box": tokens[i]["box"], "conf": tokens[i].get("conf", 0.0),
             "pred": preds[i][0], "pred_conf": preds[i][1]}
            for i in range(len(tokens))
        ]
    }
    return out

# -----------------------------
# pHash duplicate checking
# -----------------------------
def _dct2(a):  # 2D DCT via FFT (coarse)
    return np.fft.fft2(a)

def image_phash(img: Image.Image, size=32, keep=8):
    g = img.convert("L").resize((size, size), Image.BILINEAR)
    arr = np.asarray(g, dtype=np.float32)
    dct = np.real(_dct2(arr))
    dct_low = dct[:keep, :keep]
    med = np.median(dct_low)
    bits = (dct_low > med).astype(np.uint8)
    # pack to 64-bit integer
    return bits.flatten()

def hamming(a, b):
    return int(np.sum(a != b))

def load_phash_index():
    if os.path.exists(PHASH_NPY):
        idx = np.load(PHASH_NPY)
    else:
        idx = np.empty((0, 64), dtype=np.uint8)
    meta = []
    if os.path.exists(PHASH_META):
        with open(PHASH_META, "r", encoding="utf-8") as f:
            for line in f:
                meta.append(line.strip())
    return idx, meta

def save_phash_index(idx, meta):
    np.save(PHASH_NPY, idx)
    with open(PHASH_META, "w", encoding="utf-8") as f:
        for m in meta:
            f.write(m + "\n")

def check_duplicate(path):
    try:
        img = Image.open(path).convert("RGB")
    except Exception:
        return {"is_duplicate": False, "min_distance": None, "closest": None}
    bits = image_phash(img, size=32, keep=8)
    bits = bits[:64] if bits.size >= 64 else np.pad(bits, (0, 64-bits.size), constant_values=0)
    idx, meta = load_phash_index()
    if idx.shape[0] == 0:
        # first entry: add and return not duplicate
        idx_new = bits[None, :]
        save_phash_index(idx_new, [path])
        return {"is_duplicate": False, "min_distance": None, "closest": None}
    dists = [hamming(bits, idx[i]) for i in range(idx.shape[0])]
    min_d = int(np.min(dists))
    j = int(np.argmin(dists))
    # threshold ~ <=8 is very close (tweak as you like)
    is_dup = min_d <= 8
    # append new one to index (even if dup) to grow index
    idx_new = np.vstack([idx, bits[None, :]])
    meta_new = meta + [path]
    save_phash_index(idx_new, meta_new)
    return {"is_duplicate": bool(is_dup), "min_distance": min_d, "closest": meta[j]}

# -----------------------------
# Fraud scoring (simple rules)
# -----------------------------
def score_fraud(parsed):
    reasons = []
    score = 0.0

    total   = parsed.get("total")
    subtotal= parsed.get("subtotal")
    tax     = parsed.get("tax")
    items   = parsed.get("line_items", [])

    # arithmetic checks
    sum_items = sum([x.get("amount", 0.0) for x in items])
    if items and subtotal is not None:
        if abs(sum_items - subtotal) > max(0.02*max(subtotal,1.0), 0.5):
            score += 0.25
            reasons.append(f"Item sum ({sum_items:.2f}) ≠ subtotal ({subtotal:.2f})")

    if subtotal is not None and tax is not None and total is not None:
        if abs(subtotal + tax - total) > max(0.02*max(total,1.0), 0.5):
            score += 0.35
            reasons.append(f"Subtotal+Tax ({subtotal+tax:.2f}) ≠ total ({total:.2f})")
    elif total is None:
        score += 0.15
        reasons.append("Total not found")

    # date sanity (very old or clearly invalid)
    date_txt = parsed.get("date") or ""
    if date_txt and not re.search(r"\d{4}", date_txt):
        score += 0.05
        reasons.append("Suspicious date format")

    # token confidence low?
    low_conf = [t for t in parsed.get("tokens", []) if t.get("pred_conf", 1.0) < 0.35]
    if len(low_conf) >= 8:
        score += 0.1
        reasons.append("Many low-confidence tokens")

    # clamp 0..1
    score = float(max(0.0, min(1.0, score)))
    return {"fraud_score": score, "reasons": reasons}

# -----------------------------
# Overlay drawing
# -----------------------------
def draw_overlay(image_path, tokens, preds, out_dir=OVERLAY_DIR):
    try:
        im = Image.open(image_path).convert("RGB")
    except Exception:
        return None
    W,H = im.size
    draw = ImageDraw.Draw(im)
    try:
        font = ImageFont.truetype("arial.ttf", size=max(12, W//80))
    except Exception:
        font = ImageFont.load_default()

    for t, p in zip(tokens, preds):
        x0,y0,x1,y1 = t["box"]
        box = (x0*W, y0*H, x1*W, y1*H)
        lbl = p[0]
        draw.rectangle(box, outline=(255,0,0), width=2)
        draw.text((box[0]+2, box[1]-12), lbl, fill=(255,255,0), font=font)

    ts = int(time.time()*1000)
    out = os.path.join(out_dir, f"overlay_{ts}.png")
    im.save(out)
    return out

# -----------------------------
# Public API
# -----------------------------
def parse_receipt(path):
    tokens = ocr_tokens(path)
    preds  = predict_tags(tokens)
    parsed = aggregate_receipt(tokens, preds)
    dup    = check_duplicate(path)
    overlay= draw_overlay(path, tokens, preds, out_dir=OVERLAY_DIR)
    parsed["duplicate_check"] = dup
    parsed["overlay_path"]    = overlay
    return parsed

def parse_and_score(path):
    parsed = parse_receipt(path)
    fraud  = score_fraud(parsed)
    out = {"parsed": parsed, "fraud": fraud}
    return out

if __name__ == "__main__":
    # quick manual test
    sample = None
    # put a sample image under BASE_DIR for a smoke test
    for ext in IMG_EXTS:
        cand = os.path.join(BASE_DIR, "sample" + ext)
        if os.path.exists(cand):
            sample = cand; break
    if sample:
        res = parse_and_score(sample)
        print(json.dumps(res["parsed"], indent=2)[:1000], "...\n")
        print(res["fraud"])
        print("[INFO] Overlay:", res["parsed"].get("overlay_path"))
    else:
        print("[INFO] Put one sample image as C:\\Users\\sagni\\Downloads\\ReceiptRadar\\sample.jpg to test.")
```
