```python
# ReceiptRadar — accuracy graph (per-class bar chart) + confusion-matrix heatmap
# Inputs (already created by your training cell):
#   - parser_model.h5
#   - preprocessor.pkl
# Rebuilds tokens via OCR + heuristics, evaluates, and saves:
#   - accuracy_per_class.png
#   - confusion_matrix.png
#   - classification_report.txt

import os, re, json, pickle, warnings
warnings.filterwarnings("ignore")
from pathlib import Path

import numpy as np
import pandas as pd
from PIL import Image

import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow import keras
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score

# -----------------------------
# Paths
# -----------------------------
DATA_DIR   = r"C:\Users\sagni\Downloads\ReceiptRadar\archive\findit2"
OUTPUT_DIR = r"C:\Users\sagni\Downloads\ReceiptRadar"
MODEL_H5   = os.path.join(OUTPUT_DIR, "parser_model.h5")
PP_PATH    = os.path.join(OUTPUT_DIR, "preprocessor.pkl")

os.makedirs(OUTPUT_DIR, exist_ok=True)

# -----------------------------
# Load model & preprocessor
# -----------------------------
model = keras.models.load_model(MODEL_H5, compile=False)
with open(PP_PATH, "rb") as f:
    preproc = pickle.load(f)

scaler   = preproc["scaler"]
cat2id   = preproc["cat2id"]
id2cat   = preproc["id2cat"]
CATS     = [id2cat[i] for i in range(len(id2cat))]
n_classes = len(CATS)

# Regex (restore)
DATE_RE = re.compile(preproc["regex"]["date"])
PRICE_RE = re.compile(preproc["regex"]["price"], re.IGNORECASE)
TOTAL_KW = re.compile(preproc["regex"]["total_kw"], re.IGNORECASE)
TAX_KW   = re.compile(preproc["regex"]["tax_kw"], re.IGNORECASE)
ADDR_KW  = re.compile(preproc["regex"]["addr_kw"], re.IGNORECASE)

# -----------------------------
# OCR setup (match backend if possible)
# -----------------------------
OCR_BACKEND = preproc.get("ocr_backend", "none")
doctr_pred = None
easyocr_reader = None

def try_doctr():
    try:
        from doctr.models import ocr_predictor
        from doctr.io import DocumentFile
        return ocr_predictor(pretrained=True)
    except Exception:
        return None

def try_easyocr():
    try:
        import easyocr
        return easyocr.Reader(["en"], gpu=False, verbose=False)
    except Exception:
        return None

def try_tesseract():
    try:
        import pytesseract
        from pytesseract import Output
        # Windows default path (adjust if needed)
        default_tess = r"C:\Program Files\Tesseract-OCR\tesseract.exe"
        if os.path.exists(default_tess):
            pytesseract.pytesseract.tesseract_cmd = default_tess
        return True
    except Exception:
        return False

# Honor saved backend, but fall back if missing
if OCR_BACKEND == "doctr":
    doctr_pred = try_doctr()
    if doctr_pred is None:
        OCR_BACKEND = "easyocr"
if OCR_BACKEND == "easyocr":
    if easyocr_reader is None:
        easyocr_reader = try_easyocr()
    if easyocr_reader is None:
        OCR_BACKEND = "pytesseract"
if OCR_BACKEND == "pytesseract":
    if not try_tesseract():
        OCR_BACKEND = "none"
if OCR_BACKEND == "none":
    # last try: any backend available?
    doctr_pred = try_doctr()
    if doctr_pred is not None:
        OCR_BACKEND = "doctr"
    else:
        easyocr_reader = try_easyocr()
        if easyocr_reader is not None:
            OCR_BACKEND = "easyocr"
        elif try_tesseract():
            OCR_BACKEND = "pytesseract"

print(f"[INFO] Using OCR backend: {OCR_BACKEND}")

# -----------------------------
# IO helpers
# -----------------------------
IMG_EXTS = {".jpg",".jpeg",".png",".tif",".tiff",".bmp",".gif"}
PDF_EXTS = {".pdf"}

def list_docs(root):
    files = []
    for p, _, fns in os.walk(root):
        for fn in fns:
            ext = Path(fn).suffix.lower()
            if ext in IMG_EXTS or ext in PDF_EXTS:
                files.append(os.path.join(p, fn))
    return sorted(files)

# -----------------------------
# OCR wrappers
# -----------------------------
def ocr_doctr_image(img_path):
    from doctr.io import DocumentFile
    try:
        if Path(img_path).suffix.lower() in PDF_EXTS:
            doc = DocumentFile.from_pdf(img_path)
        else:
            doc = DocumentFile.from_images(img_path)
        res = doctr_pred(doc)
        tokens = []
        for p in res.pages:
            for b in p.blocks:
                for l in b.lines:
                    for wobj in l.words:
                        box = [float(wobj.geometry[0][0]), float(wobj.geometry[0][1]),
                               float(wobj.geometry[1][0]), float(wobj.geometry[1][1])]
                        txt = (wobj.value or "").strip()
                        conf = float(getattr(wobj, "confidence", 0.9) or 0.9)
                        if txt:
                            tokens.append({"text": txt, "box": box, "conf": conf})
        return tokens
    except Exception:
        return []

def ocr_easyocr_image(img_path):
    try:
        result = easyocr_reader.readtext(img_path, detail=1)
        im = Image.open(img_path).convert("RGB")
        W, H = im.size
        tokens = []
        for bbox, txt, conf in result:
            xs = [p[0] for p in bbox]; ys=[p[1] for p in bbox]
            x0,y0,x1,y1 = max(min(xs, default=0),0), max(min(ys, default=0),0), min(max(xs, default=W),W), min(max(ys, default=H),H)
            box = [x0/max(W,1), y0/max(H,1), x1/max(W,1), y1/max(H,1)]
            txt = (txt or "").strip()
            if txt:
                tokens.append({"text": txt, "box": box, "conf": float(conf)})
        return tokens
    except Exception:
        return []

def ocr_tesseract_image(img_path):
    try:
        import pytesseract
        from pytesseract import Output
        im = Image.open(img_path).convert("RGB")
        W,H = im.size
        data = pytesseract.image_to_data(im, output_type=Output.DICT)
        tokens=[]
        n = len(data.get("text",[]))
        for i in range(n):
            txt = (data["text"][i] or "").strip()
            if not txt: continue
            conf = data.get("conf",[0]*n)[i]
            try: conf = float(conf) if conf not in ("-1","") else 0.0
            except: conf = 0.0
            x,y = data["left"][i], data["top"][i]
            w,h = data["width"][i], data["height"][i]
            if W>0 and H>0 and w>0 and h>0:
                box = [x/W, y/H, (x+w)/W, (y+h)/H]
            else:
                box = [0,0,0,0]
            tokens.append({"text": txt, "box": box, "conf": conf/100.0})
        return tokens
    except Exception:
        return []

def ocr_tokens(path):
    if OCR_BACKEND == "doctr" and doctr_pred is not None:
        t = ocr_doctr_image(path)
        if t: return t
    if OCR_BACKEND == "easyocr" and easyocr_reader is not None:
        t = ocr_easyocr_image(path)
        if t: return t
    if OCR_BACKEND == "pytesseract":
        t = ocr_tesseract_image(path)
        if t: return t
    # fallback: filename tokens
    base = Path(path).stem.replace("_"," ").replace("-"," ")
    return [{"text": s, "box":[0,0,1,0.1], "conf":0.5} for s in base.split() if s]

# -----------------------------
# Features & pseudo labels (same as training)
# -----------------------------
def token_features(token):
    x0,y0,x1,y1 = token["box"]
    w = max(x1-x0, 1e-6); h = max(y1-y0, 1e-6)
    area = w*h
    cx, cy = 0.5*(x0+x1), 0.5*(y0+y1)
    txt = token["text"]
    L = len(txt)
    digits = sum(ch.isdigit() for ch in txt)
    alphas = sum(ch.isalpha() for ch in txt)
    uppers = sum(ch.isupper() for ch in txt)
    digit_ratio = digits / max(L,1)
    alpha_ratio = alphas / max(L,1)
    upper_ratio = uppers / max(alphas,1)
    price_like = 1.0 if PRICE_RE.search(txt) else 0.0
    date_like  = 1.0 if DATE_RE.search(txt) else 0.0
    tax_kw     = 1.0 if TAX_KW.search(txt) else 0.0
    total_kw   = 1.0 if TOTAL_KW.search(txt) else 0.0
    addr_kw    = 1.0 if ADDR_KW.search(txt) else 0.0
    conf       = float(token.get("conf", 0.5))
    return np.array([x0,y0,x1,y1,w,h,area,cx,cy,L,digit_ratio,alpha_ratio,upper_ratio,
                     price_like,date_like,tax_kw,total_kw,addr_kw,conf], dtype=np.float32)

def pseudo_label_tokens(tokens):
    if not tokens:
        return [], []
    feats = [token_features(t) for t in tokens]
    ys = [f[8] for f in feats]  # cy
    is_price = [bool(PRICE_RE.search(t["text"])) for t in tokens]
    has_total_kw = [bool(TOTAL_KW.search(t["text"])) for t in tokens]
    has_tax_kw   = [bool(TAX_KW.search(t["text"])) for t in tokens]
    has_date     = [bool(DATE_RE.search(t["text"])) for t in tokens]
    has_addr_kw  = [bool(ADDR_KW.search(t["text"])) for t in tokens]
    alphas_only  = [sum(ch.isalpha() for ch in t["text"]) for t in tokens]

    labels = ["other"] * len(tokens)

    # date
    for i, hd in enumerate(has_date):
        if hd: labels[i] = "date"

    # total lines by keyword proximity
    total_lines = set()
    for i, has_kw in enumerate(has_total_kw):
        if has_kw:
            cy = ys[i]
            for j in range(len(tokens)):
                if abs(ys[j]-cy) < 0.02:
                    total_lines.add(j)

    # if none, choose bottom-most price-like
    if not total_lines:
        cand = [(ys[i], i) for i in range(len(tokens)) if is_price[i]]
        if cand:
            _, idx = max(cand)
            total_lines.add(idx)

    # mark totals on those lines
    for j in total_lines:
        cy = ys[j]
        same_line = [k for k in range(len(tokens)) if abs(ys[k]-cy)<0.02]
        for k in same_line:
            if is_price[k]:
                labels[k] = "total"

    # tax lines
    for i, kw in enumerate(has_tax_kw):
        if kw:
            cy = ys[i]
            same_line = [k for k in range(len(tokens)) if abs(ys[k]-cy)<0.02]
            for k in same_line:
                if is_price[k]:
                    labels[k] = "tax"

    # merchant: longest alpha near top 15%
    top_thresh = np.quantile(ys, 0.15) if len(ys)>0 else 0.15
    top_idxs = [i for i in range(len(tokens)) if ys[i] <= top_thresh]
    if top_idxs:
        m_idx = max(top_idxs, key=lambda i: alphas_only[i])
        if alphas_only[m_idx] >= 3:
            labels[m_idx] = "merchant"

    # address
    for i in range(len(tokens)):
        if labels[i] == "other" and has_addr_kw[i]:
            labels[i] = "address"

    # subtotal (keyworded)
    min_total_y = min([ys[i] for i in range(len(tokens)) if labels[i]=="total"], default=1.0)
    for i in range(len(tokens)):
        if labels[i] == "other" and is_price[i] and ys[i] < min_total_y:
            if re.search(r"\b(sub\s*total|subtotal)\b", tokens[i]["text"], re.IGNORECASE):
                labels[i] = "subtotal"

    return labels, feats

# -----------------------------
# Build evaluation set
# -----------------------------
def build_eval(data_dir):
    files = list_docs(data_dir)
    all_X, all_y = [], []
    used = 0
    for p in files:
        toks = ocr_tokens(p)
        if not toks: continue
        labels, feats = pseudo_label_tokens(toks)
        y_ids = [cat2id.get(lbl, 0) for lbl in labels]
        all_X.extend(feats); all_y.extend(y_ids); used += 1
    return np.asarray(all_X, np.float32), np.asarray(all_y, np.int64), used

X, y, used_files = build_eval(DATA_DIR)
if X.size == 0:
    raise RuntimeError("No tokens extracted for evaluation. Ensure OCR is installed or try different receipts.")

# scale with training scaler
X_s = scaler.transform(X)

# -----------------------------
# Predict & metrics
# -----------------------------
y_prob = model.predict(X_s, verbose=0)
y_pred = y_prob.argmax(axis=1)

acc = float(accuracy_score(y, y_pred))
print(f"[INFO] Token-tagging accuracy (all tokens): {acc:.4f} on {len(y)} tokens from {used_files} file(s)")

# Classification report
rep_txt = classification_report(y, y_pred, labels=list(range(n_classes)),
                                target_names=CATS, digits=4, zero_division=0)
with open(os.path.join(OUTPUT_DIR, "classification_report.txt"), "w", encoding="utf-8") as f:
    f.write(rep_txt)

# Confusion matrix (fixed size n_classes x n_classes)
cm = confusion_matrix(y, y_pred, labels=list(range(n_classes)))

# -----------------------------
# Plot 1: Per-class accuracy bar chart
# -----------------------------
per_class_acc = []
supports = []
for c in range(n_classes):
    mask = (y == c)
    supports.append(int(mask.sum()))
    if mask.any():
        per_class_acc.append(float((y_pred[mask] == c).mean()))
    else:
        per_class_acc.append(0.0)

plt.figure(figsize=(12, 5))
xs = np.arange(n_classes)
plt.bar(xs, per_class_acc)
plt.xticks(xs, CATS, rotation=45, ha='right')
plt.ylim(0, 1.0)
plt.ylabel("Accuracy")
plt.title("Per-class Token Accuracy")
# annotate with supports
for i, (a, s) in enumerate(zip(per_class_acc, supports)):
    plt.text(i, min(a+0.02, 0.98), f"n={s}", ha="center", va="bottom", fontsize=8)
acc_path = os.path.join(OUTPUT_DIR, "accuracy_per_class.png")
plt.tight_layout()
plt.savefig(acc_path, dpi=150)
plt.close()
print(f"[INFO] Saved per-class accuracy graph -> {acc_path}")

# -----------------------------
# Plot 2: Confusion matrix heatmap (matplotlib only)
# -----------------------------
fig, ax = plt.subplots(figsize=(8, 7))
im = ax.imshow(cm, interpolation='nearest')
ax.figure.colorbar(im, ax=ax)
ax.set(
    xticks=np.arange(n_classes),
    yticks=np.arange(n_classes),
    xticklabels=CATS, yticklabels=CATS,
    ylabel='True label', xlabel='Predicted label',
    title='Confusion Matrix'
)
plt.setp(ax.get_xticklabels(), rotation=45, ha="right", rotation_mode="anchor")

# annotate cells
thresh = cm.max() / 2. if cm.max() > 0 else 0.5
for i in range(n_classes):
    for j in range(n_classes):
        val = int(cm[i, j])
        ax.text(j, i, str(val),
                ha="center", va="center",
                color="white" if val > thresh else "black", fontsize=8)

fig.tight_layout()
cm_path = os.path.join(OUTPUT_DIR, "confusion_matrix.png")
plt.savefig(cm_path, dpi=150)
plt.close()
print(f"[INFO] Saved confusion matrix heatmap -> {cm_path}")

print("[DONE]")
```
