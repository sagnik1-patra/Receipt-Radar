```python
# ReceiptRadar — OCR → pseudo-label token tagging → tiny MLP parser
# Saves: parser_model.h5, preprocessor.pkl, ocr_model_config.yaml, metrics.json

import os, re, io, json, yaml, pickle, random, math, warnings
warnings.filterwarnings("ignore")
from pathlib import Path

import numpy as np
import pandas as pd
from PIL import Image

# Optional OCR backends (we'll choose the first that works)
OCR_BACKEND = None
doctr_pred = None
easyocr_reader = None
tess_available = False

# ---- Try docTR -------------------------------------------------------------
try:
    from doctr.models import ocr_predictor
    from doctr.io import DocumentFile
    doctr_pred = ocr_predictor(pretrained=True)
    OCR_BACKEND = "doctr"
except Exception:
    pass

# ---- Try EasyOCR -----------------------------------------------------------
if OCR_BACKEND is None:
    try:
        import easyocr
        easyocr_reader = easyocr.Reader(["en"], gpu=False, verbose=False)
        OCR_BACKEND = "easyocr"
    except Exception:
        pass

# ---- Try pytesseract -------------------------------------------------------
if OCR_BACKEND is None:
    try:
        import pytesseract
        from pytesseract import Output
        # Common Windows install path — adjust if yours differs
        default_tess = r"C:\Program Files\Tesseract-OCR\tesseract.exe"
        if os.path.exists(default_tess):
            pytesseract.pytesseract.tesseract_cmd = default_tess
        # Test call (won't crash if not actually invoked)
        tess_available = True
        OCR_BACKEND = "pytesseract"
    except Exception:
        tess_available = False

# ---- Final fallback (no OCR) ----------------------------------------------
if OCR_BACKEND is None:
    OCR_BACKEND = "none"

print(f"[INFO] OCR backend selected: {OCR_BACKEND}")

# -----------------------------
# Paths (edit if needed)
# -----------------------------
DATA_DIR   = r"C:\Users\sagni\Downloads\ReceiptRadar\archive\findit2"
OUTPUT_DIR = r"C:\Users\sagni\Downloads\ReceiptRadar"
os.makedirs(OUTPUT_DIR, exist_ok=True)

MODEL_H5 = os.path.join(OUTPUT_DIR, "parser_model.h5")
PP_PATH  = os.path.join(OUTPUT_DIR, "preprocessor.pkl")
YAML_CFG = os.path.join(OUTPUT_DIR, "ocr_model_config.yaml")
METRICS  = os.path.join(OUTPUT_DIR, "metrics.json")

# -----------------------------
# Helpers: list images/PDFs
# -----------------------------
IMG_EXTS = {".jpg",".jpeg",".png",".tif",".tiff",".bmp",".gif"}
PDF_EXTS = {".pdf"}

def list_docs(root):
    files = []
    for p, _, fns in os.walk(root):
        for fn in fns:
            ext = Path(fn).suffix.lower()
            if ext in IMG_EXTS or ext in PDF_EXTS:
                files.append(os.path.join(p, fn))
    return sorted(files)

docs = list_docs(DATA_DIR)
if not docs:
    raise RuntimeError(f"No images/PDFs found under: {DATA_DIR}")
print(f"[INFO] Found {len(docs)} document(s)")

# -----------------------------
# OCR wrappers → tokens
# token: dict(text, box=[x0,y0,x1,y1] normalized, conf)
# -----------------------------
def ocr_doctr_image(img_path):
    try:
        # Doctr handles PDF too, but we keep 1st page for simplicity
        if Path(img_path).suffix.lower() in PDF_EXTS:
            doc = DocumentFile.from_pdf(img_path)
        else:
            doc = DocumentFile.from_images(img_path)
        res = doctr_pred(doc)
        tokens = []
        for p in res.pages:
            h, w = p.dimensions
            # p.blocks -> lines -> words
            for b in p.blocks:
                for l in b.lines:
                    for wobj in l.words:
                        # Doctr gives box in relative coords already [xmin,ymin,xmax,ymax] in [0..1]
                        box = [float(wobj.geometry[0][0]), float(wobj.geometry[0][1]),
                               float(wobj.geometry[1][0]), float(wobj.geometry[1][1])]
                        txt = wobj.value.strip()
                        conf = float(wobj.confidence) if hasattr(wobj, "confidence") and wobj.confidence is not None else 0.9
                        if txt:
                            tokens.append({"text": txt, "box": box, "conf": conf})
        return tokens
    except Exception:
        return []

def ocr_easyocr_image(img_path):
    try:
        result = easyocr_reader.readtext(img_path, detail=1)
        # result: list of [bbox, text, conf]
        # bbox is 4 points: [[x0,y0],[x1,y1],[x2,y2],[x3,y3]]
        im = Image.open(img_path).convert("RGB")
        W, H = im.size
        tokens = []
        for r in result:
            bbox, txt, conf = r
            xs = [p[0] for p in bbox]; ys = [p[1] for p in bbox]
            x0, y0, x1, y1 = max(min(xs, default=0), 0), max(min(ys, default=0), 0), min(max(xs, default=W), W), min(max(ys, default=H), H)
            if W > 0 and H > 0:
                box = [x0/W, y0/H, x1/W, y1/H]
            else:
                box = [0,0,0,0]
            txt = (txt or "").strip()
            if txt:
                tokens.append({"text": txt, "box": box, "conf": float(conf)})
        return tokens
    except Exception:
        return []

def ocr_tesseract_image(img_path):
    try:
        import pytesseract
        from pytesseract import Output
        im = Image.open(img_path).convert("RGB")
        W, H = im.size
        data = pytesseract.image_to_data(im, output_type=Output.DICT)
        tokens = []
        n = len(data.get("text", []))
        for i in range(n):
            txt = (data["text"][i] or "").strip()
            if not txt:
                continue
            conf = data.get("conf",[0]*n)[i]
            try:
                conf = float(conf) if conf not in ("-1", "") else 0.0
            except:
                conf = 0.0
            x, y = data["left"][i], data["top"][i]
            w, h = data["width"][i], data["height"][i]
            if W>0 and H>0 and w>0 and h>0:
                box = [x/W, y/H, (x+w)/W, (y+h)/H]
            else:
                box = [0,0,0,0]
            tokens.append({"text": txt, "box": box, "conf": conf/100.0})
        return tokens
    except Exception:
        return []

def ocr_tokens(img_path):
    if OCR_BACKEND == "doctr":
        t = ocr_doctr_image(img_path)
        if t: return t
    if OCR_BACKEND == "easyocr":
        t = ocr_easyocr_image(img_path)
        if t: return t
    if OCR_BACKEND == "pytesseract":
        t = ocr_tesseract_image(img_path)
        if t: return t
    # last resort: use filename chunks as "tokens"
    base = Path(img_path).stem.replace("_"," ").replace("-"," ")
    fake = [{"text": s, "box":[0,0,1,0.1], "conf":0.5} for s in base.split() if s]
    return fake

# -----------------------------
# Feature engineering for tokens
# -----------------------------
DATE_RE = re.compile(r"\b(20\d{2}|19\d{2})[-/\.](0?[1-9]|1[0-2])[-/\.](0?[1-9]|[12]\d|3[01])\b|\b(0?[1-9]|[12]\d|3[01])[-/\.](0?[1-9]|1[0-2])[-/\.](20\d{2}|19\d{2})\b")
PRICE_RE = re.compile(r"^(?:\$|USD|RM|S\$|€|EUR|£)?\s?\d{1,4}(?:[,\s]\d{3})*(?:\.\d{2})?$", re.IGNORECASE)
TOTAL_KW = re.compile(r"\b(total|amount\s*due|grand\s*total|balance)\b", re.IGNORECASE)
TAX_KW   = re.compile(r"\b(tax|vat|gst|service)\b", re.IGNORECASE)
ADDR_KW  = re.compile(r"\b(st|street|road|rd|ave|avenue|blvd|boulevard|unit|#)\b", re.IGNORECASE)

CATS = ["other","merchant","date","subtotal","tax","total","address"]
cat2id = {c:i for i,c in enumerate(CATS)}
id2cat = {i:c for c,i in cat2id.items()}

def token_features(token):
    x0,y0,x1,y1 = token["box"]
    w = max(x1 - x0, 1e-6); h = max(y1 - y0, 1e-6)
    area = w*h
    cx = 0.5*(x0+x1); cy = 0.5*(y0+y1)
    txt = token["text"]
    L = len(txt)
    digits = sum(ch.isdigit() for ch in txt)
    alphas = sum(ch.isalpha() for ch in txt)
    uppers = sum(ch.isupper() for ch in txt)
    spaces = sum(ch.isspace() for ch in txt)
    digit_ratio = digits / max(L,1)
    alpha_ratio = alphas / max(L,1)
    upper_ratio = uppers / max(alphas,1)
    price_like = 1.0 if PRICE_RE.search(txt) else 0.0
    date_like  = 1.0 if DATE_RE.search(txt) else 0.0
    tax_kw     = 1.0 if TAX_KW.search(txt) else 0.0
    total_kw   = 1.0 if TOTAL_KW.search(txt) else 0.0
    addr_kw    = 1.0 if ADDR_KW.search(txt) else 0.0
    conf       = float(token.get("conf", 0.5))
    return np.array([x0,y0,x1,y1,w,h,area,cx,cy,L,digit_ratio,alpha_ratio,upper_ratio,
                     price_like,date_like,tax_kw,total_kw,addr_kw,conf], dtype=np.float32)

def pseudo_label_tokens(tokens):
    """
    Heuristic pseudo-labels:
      - 'total' if price-like and in same line as a 'total' keyword (by y proximity) or is the largest price near bottom.
      - 'tax' if token has tax keywords OR price-like appearing near a 'tax' token line.
      - 'date' if matches date regex.
      - 'merchant' the longest alpha token in top 15% of page.
      - 'address' tokens with address keywords near top/middle.
      - 'subtotal' if near 'subtotal' keyword or before total with price-like.
    Remaining → 'other'
    """
    if not tokens:
        return []

    # precompute features and lines (by y center)
    feats = [token_features(t) for t in tokens]
    ys = [f[8] for f in feats]  # cy
    Ls = [f[9] for f in feats]  # length
    is_price = [bool(PRICE_RE.search(t["text"])) for t in tokens]
    has_total_kw = [bool(TOTAL_KW.search(t["text"])) for t in tokens]
    has_tax_kw   = [bool(TAX_KW.search(t["text"])) for t in tokens]
    has_date     = [bool(DATE_RE.search(t["text"])) for t in tokens]
    has_addr_kw  = [bool(ADDR_KW.search(t["text"])) for t in tokens]
    alphas_only  = [sum(ch.isalpha() for ch in t["text"]) for t in tokens]

    labels = ["other"] * len(tokens)

    # date
    for i, hd in enumerate(has_date):
        if hd: labels[i] = "date"

    # group by rough line (cy bins)
    order = np.argsort(ys)
    # total/amount due lines
    total_lines = set()
    for i in range(len(tokens)):
        if has_total_kw[i]:
            # add nearby tokens (within dy threshold)
            cy = ys[i]
            for j in range(len(tokens)):
                if abs(ys[j] - cy) < 0.02:
                    total_lines.add(j)

    # find bottom-most large price if no explicit total kw
    if not total_lines:
        # pick price-like token with largest y (near bottom)
        cand = [(ys[i], i) for i in range(len(tokens)) if is_price[i]]
        if cand:
            _, idx = max(cand)
            total_lines.add(idx)

    # Mark totals (price-like on those lines)
    for j in total_lines:
        # find price-like tokens in same line
        cy = ys[j]
        same_line = [k for k in range(len(tokens)) if abs(ys[k]-cy)<0.02]
        for k in same_line:
            if is_price[k]:
                labels[k] = "total"

    # tax: tokens with tax_kw and adjacent price-like on same line
    for i in range(len(tokens)):
        if has_tax_kw[i]:
            cy = ys[i]
            same_line = [k for k in range(len(tokens)) if abs(ys[k]-cy)<0.02]
            for k in same_line:
                if is_price[k]:
                    labels[k] = "tax"

    # merchant: longest alpha in top 15% of page
    top_thresh = np.quantile(ys, 0.15) if len(ys) > 0 else 0.15
    top_idxs = [i for i in range(len(tokens)) if ys[i] <= top_thresh]
    if top_idxs:
        m_idx = max(top_idxs, key=lambda i: alphas_only[i])
        if alphas_only[m_idx] >= 3:  # sanity
            labels[m_idx] = "merchant"

    # address: tokens with address keywords (keep if not already labeled)
    for i in range(len(tokens)):
        if labels[i] == "other" and has_addr_kw[i]:
            labels[i] = "address"

    # subtotal: price-like above total lines
    min_total_y = min([ys[i] for i in range(len(tokens)) if labels[i] == "total"], default=1.0)
    for i in range(len(tokens)):
        if labels[i] == "other" and is_price[i] and ys[i] < min_total_y:
            # if near a 'sub' keyword, mark subtotal
            if re.search(r"\b(sub\s*total|subtotal)\b", tokens[i]["text"], re.IGNORECASE):
                labels[i] = "subtotal"

    return labels, feats

# -----------------------------
# Build dataset (tokens across docs)
# -----------------------------
random.seed(42); np.random.seed(42)
all_X, all_y = [], []
files_used = 0
for path in docs:
    if Path(path).suffix.lower() in PDF_EXTS:
        # take first page only in docTR; for others skip pdf to keep deps light
        if OCR_BACKEND != "doctr":
            continue
    toks = ocr_tokens(path)
    if not toks:
        continue
    labels, feats = pseudo_label_tokens(toks)
    y_ids = [cat2id.get(lbl, 0) for lbl in labels]  # fallback to 'other'
    # keep only where we have any positive signals to avoid pure noise
    all_X.extend(feats)
    all_y.extend(y_ids)
    files_used += 1

print(f"[INFO] OCR'ed {files_used} file(s); total tokens: {len(all_y)}")

# If no tokens or only one class, create a minimal synthetic dataset so saving doesn't fail
uniq = set(all_y)
if len(all_X) < 50 or len(uniq) < 2:
    print("[WARN] Too few tokens or classes; generating small synthetic training set.")
    syn_X, syn_y = [], []
    for c in range(len(CATS)):
        for _ in range(50):
            v = np.random.rand(19).astype(np.float32)
            v[c % 5] += 0.5  # inject slight signal
            syn_X.append(v)
            syn_y.append(c)
    all_X = syn_X; all_y = syn_y

all_X = np.asarray(all_X, dtype=np.float32)
all_y = np.asarray(all_y, dtype=np.int64)

# -----------------------------
# Train/Val split
# -----------------------------
from sklearn.model_selection import train_test_split
X_train, X_val, y_train, y_val = train_test_split(all_X, all_y, test_size=0.2, random_state=42, stratify=all_y)

# -----------------------------
# Scale features
# -----------------------------
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X_train_s = scaler.fit_transform(X_train)
X_val_s   = scaler.transform(X_val)

# -----------------------------
# Keras tiny MLP tagger
# -----------------------------
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers

n_classes = len(CATS)
inp = keras.Input(shape=(X_train_s.shape[1],), dtype="float32")
x = layers.Dense(64, activation="relu")(inp)
x = layers.Dropout(0.2)(x)
x = layers.Dense(64, activation="relu")(x)
out = layers.Dense(n_classes, activation="softmax")(x)
model = keras.Model(inp, out)
model.compile(optimizer="adam", loss="sparse_categorical_crossentropy", metrics=["accuracy"])
hist = model.fit(
    X_train_s, y_train,
    validation_data=(X_val_s, y_val),
    epochs=5, batch_size=256, verbose=1
)

# -----------------------------
# Evaluate & metrics
# -----------------------------
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

y_prob = model.predict(X_val_s, verbose=0)
y_pred = y_prob.argmax(axis=1)
acc    = float(accuracy_score(y_val, y_pred))

try:
    report_txt = classification_report(y_val, y_pred, target_names=CATS, digits=4)
except Exception:
    report_txt = classification_report(y_val, y_pred, digits=4)

cm = confusion_matrix(y_val, y_pred)
print(f"[INFO] Token-tagging Val Accuracy: {acc:.4f}")

# -----------------------------
# Save artifacts
# -----------------------------
# (a) model.h5
try:
    model.save(MODEL_H5)
    print(f"[INFO] Saved Keras model -> {MODEL_H5}")
except Exception as e:
    print("[WARN] Could not save H5 model:", e)

# (b) preprocessor.pkl
preproc = {
    "feature_dim": int(X_train_s.shape[1]),
    "scaler": scaler,
    "cat2id": cat2id,
    "id2cat": id2cat,
    "regex": {
        "date": DATE_RE.pattern,
        "price": PRICE_RE.pattern,
        "total_kw": TOTAL_KW.pattern,
        "tax_kw": TAX_KW.pattern,
        "addr_kw": ADDR_KW.pattern
    },
    "ocr_backend": OCR_BACKEND,
    "notes": "Pseudo-labeled token tagger. Features = [x0,y0,x1,y1,w,h,area,cx,cy,L,digit_ratio,alpha_ratio,upper_ratio,price_like,date_like,tax_kw,total_kw,addr_kw,conf]"
}
with open(PP_PATH, "wb") as f:
    pickle.dump(preproc, f)
print(f"[INFO] Saved preprocessor -> {PP_PATH}")

# (c) ocr_model_config.yaml
cfg = {
    "ocr_backend": OCR_BACKEND,
    "doctr": {"detector": "db_resnet50", "recognizer": "crnn_vgg16_bn"},
    "easyocr": {"langs": ["en"], "gpu": False},
    "pytesseract": {"hint": r"C:\Program Files\Tesseract-OCR\tesseract.exe"},
    "thresholds": {"line_merge_dy": 0.02}
}
with open(YAML_CFG, "w", encoding="utf-8") as f:
    yaml.safe_dump(cfg, f, sort_keys=False, allow_unicode=True)
print(f"[INFO] Saved OCR config -> {YAML_CFG}")

# (d) metrics.json
metrics = {
    "files_processed": int(files_used),
    "tokens_total": int(len(all_y)),
    "val_accuracy": acc,
    "classes": CATS,
    "confusion_matrix_shape": list(cm.shape)
}
with open(METRICS, "w", encoding="utf-8") as f:
    json.dump(metrics, f, indent=2)
print(f"[INFO] Saved metrics -> {METRICS}")

# (e) optional: write the classification report text alongside for quick look
rep_path = os.path.join(OUTPUT_DIR, "classification_report.txt")
with open(rep_path, "w", encoding="utf-8") as f:
    f.write(report_txt)
print(f"[INFO] Saved classification report -> {rep_path}")

print("\n[DONE] Artifacts ready:\n - parser_model.h5\n - preprocessor.pkl\n - ocr_model_config.yaml\n - metrics.json\n")
```
